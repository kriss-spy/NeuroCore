# 03 Baseline MLP Training
Train a simple MLP on a random 10% subset of episodes and evaluate against the remaining episodes.

```python
import os
import sys
sys.path.append(os.path.abspath('..'))

import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

from src.baseline import BaselineConfig, run_baseline

feature_path = os.path.abspath('../results/features_resnet18.pt')
```

## 1. Train Baseline Model
We keep the architecture and hyperparameters consistent with the plan.

```python
config = BaselineConfig()
metrics = run_baseline(feature_path=feature_path, config=config)

print(f"Baseline MSE: {metrics['mse']:.6f}")
print("Per-joint MSE:", metrics["per_joint_mse"])
print("Train episodes:", metrics["train_episodes"])
```

## 2. Loss Curves
Plot training and validation loss curves for quick sanity check.

```python
plt.figure(figsize=(8, 4))
plt.plot(metrics["train_losses"], label="Train")
plt.plot(metrics["val_losses"], label="Validation")
plt.title("Baseline MLP Loss Curves")
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.legend()
plt.show()
```

## 3. Per-Joint MSE
Inspect which joints are most challenging to predict.

```python
per_joint = np.array(metrics["per_joint_mse"])
plt.figure(figsize=(8, 4))
sns.barplot(x=list(range(len(per_joint))), y=per_joint)
plt.title("Per-Joint MSE (Baseline)")
plt.xlabel("Joint Index")
plt.ylabel("MSE")
plt.show()
```
